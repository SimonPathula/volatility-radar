```python
economic['date'] = pd.to_datetime(economic['date'])
price_combined['timestamp'] = pd.to_datetime(price_combined['timestamp'])

full_merged = pd.merge(economic, price_combined, left_on='date', right_on='timestamp', how='left')
full_merged.drop(columns=['timestamp'], inplace=True)

print(full_merged.shape)
print(full_merged.head(3))
```

```python
def map_pair(currency):
    mapping = {
        'EUR': 'EURUSD',
        'GBP': 'GBPUSD',
        'JPY': 'USDJPY',
        'USD': None  # USD affects all pairs
    }
    return mapping.get(currency)

economic['pair'] = economic['currency'].map(map_pair)
```

```python
# map non-USD currencies directly to their pair
eur_events = economic[economic['currency'] == 'EUR'].copy()
eur_events['pair'] = 'EURUSD'

gbp_events = economic[economic['currency'] == 'GBP'].copy()
gbp_events['pair'] = 'GBPUSD'

jpy_events = economic[economic['currency'] == 'JPY'].copy()
jpy_events['pair'] = 'USDJPY'

# USD events go to all three pairs
usd_events = economic[economic['currency'] == 'USD'].copy()

usd_eurusd = usd_events.copy()
usd_eurusd['pair'] = 'EURUSD'

usd_gbpusd = usd_events.copy()
usd_gbpusd['pair'] = 'GBPUSD'

usd_usdjpy = usd_events.copy()
usd_usdjpy['pair'] = 'USDJPY'

# stack everything
economic_mapped = pd.concat([eur_events, gbp_events, jpy_events, usd_eurusd, usd_gbpusd, usd_usdjpy], ignore_index=True)

print(economic_mapped.shape)
print(economic_mapped['pair'].value_counts())
```

```python
price_combined['timestamp'] = pd.to_datetime(price_combined['timestamp'])
economic_mapped['date'] = pd.to_datetime(economic_mapped['date'])

full_merged = pd.merge(economic_mapped, price_combined, left_on=['date', 'pair'], right_on=['timestamp', 'pair'], how='left')
full_merged.drop(columns=['timestamp'], inplace=True)

print(full_merged.shape)
print(full_merged.head(3))
```

```python
print(full_merged.isnull().sum())
```

```python
full_merged.dropna(subset=['open', 'high', 'low', 'close'], inplace=True)
print(full_merged.shape)
print(full_merged.isnull().sum())
```
